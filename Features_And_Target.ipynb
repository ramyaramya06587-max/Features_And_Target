```python
# TASK 31: FEATURES AND TARGET SELECTION
# Veda Technology - AI & ML Internship

import pandas as pd
import numpy as np

# 1. Load Dataset
df = pd.read_csv("iris.csv", sep=None, engine="python")

# Remove extra spaces from column names
df.columns = df.columns.str.strip().str.lower()

print("Dataset Loaded Successfully!")
print("\nDataset Preview:")
print(df.head())

# 2. Dataset Information
print("\nDataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())

# 3. Select Input Features (X)
X = df[
    ["sepal_length", "sepal_width",
     "petal_length", "petal_width"]
]

print("\n========== INPUT FEATURES (X) ==========")
print(X.head())

# 4. Select Target Variable (y)
y = df["species"]

print("\n========== TARGET VARIABLE (y) ==========")
print(y.head())

# 5. Display Target Classes
print("\nTarget Classes:")
print(y.unique())

# 6. Feature and Target Details
print("\n========== SELECTION DETAILS ==========")
print("Feature Columns:", X.columns.tolist())
print("Target Column:", y.name)
print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)

# 7. Check Missing Values
print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())

# 8. Reason for Selection
print("\n========== REASON FOR SELECTION ==========")

print("""
Input Features (X):
Sepal Length, Sepal Width,
Petal Length and Petal Width.

These measurements help identify flower species.

Target Variable (y):
Species.

The species column represents the flower category
that the machine learning model needs to predict.
""")

# 9. Conclusion
print("\n========== CONCLUSION ==========")
print("""
Four input features and one target variable
were successfully selected.

Task 31 completed successfully!
""")
```
